```python
# Cell 2 — Authenticate + Download
import os
from kaggle_secrets import UserSecretsClient
from huggingface_hub import snapshot_download, login

# Get token from Kaggle secrets
secrets = UserSecretsClient()
hf_token = secrets.get_secret("HF_TOKEN")

# Login to HuggingFace
login(token=hf_token)
print("✅ Logged in to HuggingFace")

# Download model — try 8B first
MODEL_REPO = "Qwen/Qwen3-Embedding-8B"   # if this fails, try "Qwen/Qwen3-Embedding"

snapshot_download(
    repo_id=MODEL_REPO,
    local_dir="/kaggle/working/qwen3-embedding",
    token=hf_token,
    ignore_patterns=["*.msgpack", "*.h5", "flax_*", "*.ot", "rust_model.ot"]
)

print("✅ Download complete")
!du -sh /kaggle/working/qwen3-embedding
```
